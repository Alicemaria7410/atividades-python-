```python
#4 questão:
# Entrada dos 10 números
numeros = [int(input(f"Número {i+1}: ")) for i in range(10)]

# Contagem da maior sequência crescente
maior_seq = seq_atual = 1

for i in range(1, len(numeros)):
    if numeros[i] > numeros[i-1]:
        seq_atual += 1
        maior_seq = max(maior_seq, seq_atual)
    else:
        seq_atual = 1

print("\nLista:", numeros)
print("Maior sequência crescente:", maior_seq)
```

```python
# 5 questão:
pontos = [(7, 3), (12, -4), (0,-2), (-10,-25), (67, 50), (0, 0)]
q1 = q2 = q3 = q4 = eixos = 0
maior_ponto, maior_val = None, -1

for x, y in pontos:
    if x > 0 and y > 0: q1 += 1
    elif x < 0 and y > 0: q2 += 1
    elif x < 0 and y < 0: q3 += 1
    elif x > 0 and y < 0: q4 += 1
    else: eixos += 1

    if (x**2 + y**2) > maior_val:
        maior_val = x**2 + y**2
        maior_ponto = (x, y)

print(f"Q1:{q1} Q2:{q2} Q3:{q3} Q4:{q4} Eixos:{eixos}")
print(f"Maior ponto: {maior_ponto}")
```

```python
#6 questão:
# 1 e 2. Ler palavras e montar o dicionário
freq = {}
for _ in range(10):
    p = input("Palavra: ").strip().lower()
    freq[p] = freq.get(p, 0) + 1

# 3. Achar o maior valor sem max()
maior = 0
for v in freq.values():
    if v > maior: maior = v

# 4. Achar os empates e mostrar resultados
empates = [p for p, v in freq.items() if v == maior]

print("Dicionário:", freq)
print("Maior frequência:", maior)
print("Empates:", empates)
```

```python
#7 questão:
turma = []
for _ in range(6):
    nome = input("Nome: ")
    m = (float(input("N1: ")) + float(input("N2: ")) + float(input("N3: "))) / 3
    turma.append({"nome": nome, "media": m})

mg = sum(a["media"] for a in turma) / 6
maior = sorted(a["media"] for a in turma)[-1]

print(f"\nMédia Geral: {mg:.2f}")
print(f"Alunos >= Média Geral: {sum(1 for a in turma if a['media'] >= mg)}")
print(f"Maior Média ({maior:.2f}): {', '.join(a['nome'] for a in turma if a['media'] == maior)}")
```

```python
#8 questão:
jogadores = [{"nome": "A", "p": [10, 20, 15]}, {"nome": "B", "p": [30, 25, 35]}, {"nome": "C", "p": [12, 14, 13]}, {"nome": "D", "p": [40, 10, 20]}, {"nome": "E", "p": [30, 25, 35]}]
res = []

for j in jogadores:
    p = j["p"]
    tot = p[0] + p[1] + p[2]
    maior = p[0] if p[0] > p[1] and p[0] > p[2] else (p[1] if p[1] > p[2] else p[2])
    menor = p[0] if p[0] < p[1] and p[0] < p[2] else (p[1] if p[1] < p[2] else p[2])
    res.append({"nome": j["nome"], "tot": tot, "dif": maior - menor})
    print(f"{j['nome']}: Total = {tot} | Média = {tot/3:.1f}")

print("\nMaior Total:", [r["nome"] for r in res if r["tot"] == max([x["tot"] for x in res])])
print("Mais Regular:", [r["nome"] for r in res if r["dif"] == min([x["dif"] for x in res])])
```

```python
#9 questão:
jgs = [{"n": "Zayan", "p": [10,20,15]}, {"n": "Ayla", "p": [30,25,30]}, {"n": "Rafael", "p": [5,5,5]}, {"n": "Yan", "p": [20,20,20]}, {"n": " Mel", "p": [10,30,20]}]
m_tot, m_dif = -1, 999

for j in jgs:
    p = j["p"]
    j["t"], j["m"] = p[0]+p[1]+p[2], (p[0]+p[1]+p[2])/3
    maior = p[0] if p[0]>=p[1] and p[0]>=p[2] else (p[1] if p[1]>=p[2] else p[2])
    menor = p[0] if p[0]<=p[1] and p[0]<=p[2] else (p[1] if p[1]<=p[2] else p[2])
    j["d"] = maior - menor
    if j["t"] > m_tot: m_tot = j["t"]
    if j["d"] < m_dif: m_dif = j["d"]

print("--- RESULTADOS ---", *[f"{j['n']} | Total: {j['t']} | Média: {j['m']:.1f}" for j in jgs], sep="\n")
print(f"Maior Total ({m_tot}):", [j['n'] for j in jgs if j['t'] == m_tot])
print(f"Mais Regular ({m_dif}):", [j['n'] for j in jgs if j['d'] == m_dif])
```

```python
# 10 questão:
agenda = []

# Cadastro
while (nome := input("Nome cadastrar (FIM p/ sair): ")) != "FIM":
    if any(c["nome"] == nome for c in agenda):
        print("Já cadastrado!")
    else:
        agenda.append({"nome": nome, "telefone": input("Telefone: ")})

# Consulta
while (busca := input("Nome consultar (FIM p/ sair): ")) != "FIM":
    resultado = [c["telefone"] for c in agenda if c["nome"] == busca]
    print(f"Telefone: {resultado[0]}" if resultado else "Não encontrado.")
```

```python
#11 questão:
votos = {1:5, 6: 0, 3: 0, 4: 0}
total = 0

# Entrada dos votos
while (voto := int(input("Voto (1-4) ou 0 p/ sair: "))) != 0:
    if voto in votos:
        votos[voto] += 1
        total += 1
    else:
        print("Inválido!")

# Percentuais
for op, qtd in votos.items():
    print(f"Opção {op}: {qtd} votos ({(qtd/total*100 if total else 0):.1f}%)")

# Maior voto sem max() e busca de vencedores com empate em lista
maior = -1
for qtd in votos.values():
    if qtd > maior: maior = qtd

vencedores = [str(op) for op, qtd in votos.items() if qtd == maior > 0]
print("Vencedores:", ", ".join(vencedores) if vencedores else "Nenhum")
```

```python
#12 questão:
estoque = {101: 50, 102: 30}
historico = []

 8 movimentações em tuplas (codigo, tipo, quantidade)
movimentacoes = [
    (101, "E", 20),
    (102, "S", 10),
    (101, "S", 100),  # Recusada
    (102, "E", 5),
    (101, "S", 10),
    (103, "E", 15),
    (102, "S", 50),  # Recusada
    (103, "S", 5),
]

for cod, tipo, qtd in movimentacoes:
    atual = estoque.get(cod, 0)

    if tipo == "S" and qtd > atual:
        status = "Recusado"
    else:
        status = "Aprovado"
        estoque[cod] = atual + qtd if tipo == "E" else atual - qtd

    historico.append((cod, tipo, qtd, status))

print("Histórico:", historico)
print("Estoque final:", estoque)
```

```python
# 13 questão:
vendedores = []
maior = 0
total_geral = 0

# 1. Leitura e Cálculo dos Totais
for i in range(5):
    nome = input("Nome: ")
    vendas = [float(input(f"Venda {j+1}: ")) for j in range(4)]

    total = 0
    for v in vendas:
        total += v

    if total > maior:
        maior = total

    total_geral += total
    vendedores.append({"nome": nome, "total": total})

media = total_geral / 5

# 2. Saída dos Resultados
print("\n--- RESULTADO ---")
acima_media = 0
for v in vendedores:
    print(f"{v['nome']}: R$ {v['total']:.2f}")
    if v["total"] > media:
        acima_media += 1

print(f"Total Geral: R$ {total_geral:.2f}")
print(f"Média: R$ {media:.2f}")
print(f"Maior valor: R$ {maior:.2f}")
print(f"Acima da média: {acima_media}")
```

```python
#14 questão:
produtos = [{"codigo": 1, "nome": "Camiseta", "preco": 50.0, "estoque": 10}, {"codigo": 2, "nome": "Calça", "preco": 100.0, "estoque": 5}, {"codigo": 3, "nome": "Tênis", "preco": 200.0, "estoque": 3}, {"codigo": 4, "nome": "Boné", "preco": 30.0, "estoque": 8}, {"codigo": 5, "nome": "Meia", "preco": 15.0, "estoque": 2}]
pedidos = [(1, 2), (3, 5), (5, 2), (5, 1), (9, 1), (2, 3)]
total = 0
for c, q in pedidos:
    p = next((x for x in produtos if x["codigo"] == c), None)
    if p and p["estoque"] >= q:
        p["estoque"] -= q; total += p["preco"] * q
        print(f"ATENDIDO: {q}x {p['nome']}")
    else: print(f"RECUSADO: Código {c}")
print(f"Total: R$ {total:.2f}")
```

```python
#15 questão:
produtos = [
    {"cod": 1, "nome": "Arroz", "preco": 20.0, "est": 10, "v": 0},
    {"cod": 2, "nome": "Feijão", "preco": 8.0, "est": 15, "v": 0},
    {"cod": 3, "nome": "carne", "preco":50.0, "est": 5, "v": 0},
    {"cod": 4, "nome": "balinha", "preco": 4.0, "est": 12, "v": 0},
    {"cod": 5, "nome": "Cafe", "preco": 14.0, "est": 8, "v": 0}
]
hist = []

while (c := int(input("Cód (1-5 ou 0): "))) != 0:
    p = next((x for x in produtos if x["cod"] == c), None)
    if p and 0 < (q := int(input(f"Qtd ({p['nome']}): "))) <= p["est"]:
        p["est"] -= q; p["v"] += q
        hist.append((c, p["nome"], q, q * p["preco"]))

print("\n--- RELATÓRIO ---")
for p in produtos: print(f"{p['nome']} | Resta: {p['est']} | Vendidos: {p['v']}")

mv = max(p["v"] * p["preco"] for p in produtos)
print("Maior valor:", [p["nome"] for p in produtos if p["v"] * p["preco"] == mv > 0])
print("Sem venda:", [p["nome"] for p in produtos if p["v"] == 0])
print("Histórico:", hist)
print("Tupla-Resumo:", (len(hist), sum(p["v"] for p in produtos), sum(p["v"] * p["preco"] for p in produtos)))
```
