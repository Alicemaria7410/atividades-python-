```python
# 6 questão:
# Lista para guardar os nomes
nomes = [] # Inicializa a lista de nomes

#Laço para leitura dos nomes,até digitar o fim
while True:
    nome = input("Digite um nome: ")

    if nome.upper() == "FIM":
        break  # Para a leitura

    if nome not in nomes:
        nomes.append(nome)  # Adiciona só se não for rep

print("Lista de nomes:", nomes)
print("Quantidade de nomes diferentes:", len(nomes))
```

```python
#7 questão:
#Criação da tupla com os dias da semana (índices de 0 a 6)
dias = ("Domingo", "Segunda-feira", "Terça-feira", "Quarta-feira", "Quinta-feira", "Sexta-feira", "Sábado")

# Leitura do número digitado
numero = int(input("Digite um número de 1 a 7: "))

 # Validação da entrada e exibição do resultado
if 1 <= numero <= 7:
    # Subtrai 1 porque os índices em Python começam em 0
    print(f"O dia correspondente é: {dias[numero - 1]}")
else:
    print("Opção inválida.")
```

```python
#9 questão:

# Criação do dicionário com os dados iniciais do aluno
aluno = {
    'nome': 'Alice',
    'turma': '7A',
    'nota1': 9.0,
    'nota2': 10.0
}

# Cálculo da média e adição ao dicionário
aluno['media'] = (aluno['nota1'] + aluno['nota2']) / 2

#Definição da situação com base nas regras do enunciado
if aluno['media'] >= 7.0:
    aluno['situacao'] = 'aprovado'
elif aluno['media'] >= 5.0:
    aluno['situacao'] = 'recuperação'
else:
    aluno['situacao'] = 'reprovado'

# Exibição dos dados do aluno (Teste executado)
for chave, valor in aluno.items():
    print(f"{chave.capitalize()}: {valor}")
```

```python
#11 questão:
#Cria um dicionário vazio para guardar as palavras e quantidades
contagem = {}

#Lê 8 palavras do usuário usando um laço de repetição
for i in range(8):
    palavra = input(f"Digite a palavra {i+1}: ").strip().lower()

    #Verifica se a palavra já está no dicionário
    if palavra in contagem:
        contagem[palavra] += 1  # Incrementa se já existir
    else:
        contagem[palavra] = 1   # Cria com valor 1 se não existir

#Exibe o resultado final
print("\n--- Frequência das Palavras ---")
for palavra, qtd in contagem.items():
    print(f"{palavra}: {qtd}")
```

```python
#12 questão:

estoque = {"morango": 10, "banana": 5, "laranja": 8, "uva": 12}

p = input("Produto: ").lower()
q = int(input("Quantidade: "))

if p in estoque and estoque[p] >= q:
    estoque[p] -= q
    print("Venda feita!")
else:
    print("Erro no produto ou quantidade insuficiente!")

print(estoque)
```

```python
#13 questão:
turma = []

for i in range(5):
    nome = input("Nome do aluno: ")
    if nome.upper() == "FIM":
        break

    n1_str = input("Nota 1: ")
    if n1_str.upper() == "FIM":
        break
    n1 = float(n1_str)

    n2_str = input("Nota 2: ")
    if n2_str.upper() == "FIM":
        break
    n2 = float(n2_str)

    media = (n1 + n2) / 2

    # Cria o dicionário e adiciona na lista
    aluno = {"nome": nome, "notas": [n1, n2], "media": media}
    turma.append(aluno)

# Exibe o resultado
print("\n--- RESULTADOS ---")
for aluno in turma:
    print(f"Aluno: {aluno['nome']} | Média: {aluno['media']:.1f}")
```

```python
#14 questão:
#Lista com as 5 tuplas (codigo, nome, preco)
produtos = [(101, "Arroz", 25.5), (102, "Feijão", 8.9), (103, "Café", 14.0), (104, "Leite", 5.2), (105, "Açúcar", 4.5)]

#Leitura do código
codigo_busca = int(input("Digite o código: "))

#Busca e exibição
for cod, nome, preco in produtos:
    if cod == codigo_busca:
        print(f"{nome} - R$ {preco}")
        break
else:
    print("Código não encontrado.")
```

```python
#15 questão:
#Cadastro com Lista e Dicionários
estoque = [
    {"codigo": 1, "nome": "Tablet", "quantidade": 10, "preco": 50.0},
    {"codigo": 2, "nome": "Caneta digital", "quantidade": 15, "preco": 30.0},
    {"codigo": 3, "nome": "", "quantidade": 5, "preco": 800.0},
    {"codigo": 4, "nome": "Fone", "quantidade": 8, "preco": 100.0},
    {"codigo": 5, "nome": "Webcam", "quantidade": 12, "preco": 150.0}
]

# (a) Mostrar itens
print("Itens:", estoque)

# (b) e (c) Total e maior valor (sem max)
total = 0
maior = estoque[0]

for item in estoque:
    valor = item["quantidade"] * item["preco"]
    total += valor
    if valor > (maior["quantidade"] * maior["preco"]):
        maior = item

print(f"Valor total: R$ {total:.2f}")
print("Item de maior valor:", maior["nome"])

# (d) Tupla resumo
tupla_resumo = (len(estoque), total)
print("Resumo:", tupla_resumo)
```
