```python
#8 questão:
anterior = int(input("Digite o 1º número: "))
repetido = False
for i in range(2, 9):
    atual = int(input(f"Digite o {i}º número: "))
    if atual == anterior:
     repetido = True
    anterior = atual

if repetido:
    print("Existem valores consecutivos iguais.")
else:
    print("Não existem valores consecutivos iguais.")
```

```python
# Questão 9:
saldo = 500.0

for i in range(6):
    tipo = int(input("Tipo (1 = Depósito, 2 = Saque): "))
    valor = float(input("Valor: "))

    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo final: R$ {saldo:.2f}")
```

```python
#11 questão:
n = int(input("Digite um número inteiro maior que 1: "))
primo = True

if n <= 1:
    primo = False
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            primo = False
            break

if primo:
    print(f"O número {n} é primo.")
else:
    print(f"O número {n} não é primo.")
```

O código descobre se um número é primo, fazendo os passos explicados abaixo:Leitura: Recebe o número digitado pelo usuário.
​Teste: Se o número for menor ou igual a 1, define como não primo; se for maior, testa se ele divide de forma exata por algum número até a sua raiz quadrada.
​Resultado: Mostra na tela se o número é primo ou não.

```python
#12 questão:
n = int(input("Digite um número positivo N: "))
total_divisores = 0

print(f"Divisores de {n}:")
for i in range(1, n + 1):
    if n % i == 0:
        print(i, end=" ")
        total_divisores += 1

print(f"\nQuantidade de divisores: {total_divisores}")
```

O código serve para contar os divisores de um número.
 Você digita um número, ele testa a divisão por todos os números de 1 até ele e exibe na tela apenas os que dividem de forma exata (resto zero). No final, ele mostra quais são esses divisores e quantos foram encontrados no total.

```python
#13 questão:
a = int(input("Digite o valor de A: "))
b = int(input("Digite o valor de B (A < B): "))
qtd_primos = 0

print(f"Primos no intervalo [{a}, {b}]:")
for num in range(a, b + 1):
    if num > 1:
        eh_primo = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                eh_primo = False
                break
        if eh_primo:
            print(num, end=" ")
            qtd_primos += 1

print(f"\nTotal de números primos encontrados: {qtd_primos}")
```

O código encontra e conta os números primos dentro de um intervalo escolhido pelo usuário.
​Entrada: Recebe um valor inicial e outro final para definir a faixa de busca.
​Processamento: Checa cada número do intervalo, confirmando se ele só divide por 1 e por ele mesmo.
​Saída: Mostra os primos encontrados na tela e exibe a quantidade total no

```python
#14 questão:
secreto = 37
tentativas = 5
acertou = False

for i in range(1, tentativas + 1):
    palpite = int(input(f"Tentativa {i}/{tentativas}. Digite seu palpite: "))

    if palpite == secreto:
        print(" Você acertou!Parabéns!")
        acertou = True
        break
    elif palpite < secreto:
        print("O número secreto é maior.")
    else:
        print("O número secreto é menor.")

if not acertou:
    print(f"Suas tentativas acabaram! O número secreto era {secreto}.")
```

O código cria um jogo de adivinhação em que o jogador tem 5 chances para acertar o número 37.
​O sistema pede um palpite a cada rodada e diz se o número correto é maior ou menor.
​Se você acertar, o jogo parabeniza e encerra.
​Se errar 5 vezes, ele avisa que as tentativas acabaram e revela a resposta.

```python
#15 questão
saldo, entradas, saidas, recusadas, total_in, total_out = 200.0, 0, 0, 0, 0.0, 0.0

while (op := int(input("Código (1=Entrada, 2=Retirada, 0=Sair): "))) != 0:
    if op in (1, 2):
        v = float(input("Valor: "))
        if op == 1:
            saldo += v; total_in += v; entradas += 1
        elif v <= saldo:
            saldo -= v; total_out += v; saidas += 1
        else:
            print("Operação RECUSADA: Saldo insuficiente.")
            recusadas += 1

print(f"\n--- RESUMO DO DIA ---\nSaldo final: R$ {saldo:.2f}\nEntradas: {entradas} (Total: R$ {total_in:.2f})\nRetiradas: {saidas} (Total: R$ {total_out:.2f})\nRecusadas: {recusadas}")
```
