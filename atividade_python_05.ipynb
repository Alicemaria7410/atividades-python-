5 Questão

➥​Exceção (ValueError): Ocorre na leitura das notas (float(input(...))) quando o usuário digita textos, letras ou símbolos não numéricos em vez de um número válido.  
​
➥Estratégia: Envolver a conversão das entradas em um bloco try-except ValueError (com ou sem um laço while) para capturar o erro e pedir novamente uma entrada válida ao usuário.

```python
# 6 Questão:

 # Função para ler um número inteiro e refazer a leitura se houver erro
def ler_inteiro(mensagem):
    while True:
        try:
            return int(input(mensagem))
        except ValueError:
            print("Entrada inválida! Digite apenas números inteiros.")

# Leitura dos três números
num1 = ler_inteiro("Digite o primeiro número: ")
num2 = ler_inteiro("Digite o segundo número: ")
num3 = ler_inteiro("Digite o terceiro número: ")

# Lógica para encontrar o maior sem usar max()
maior = num1
if num2 > maior:
    maior = num2
if num3 > maior:
    maior = num3

print(f"O maior número é: {maior}")
```

Questão 6

➥​Exceção (ValueError): Ocorre quando o usuário digita um valor que não pode ser convertido para número inteiro (como texto ou símbolos).

➥​Estratégia: Utiliza um laço de repetição while True em conjunto com o bloco try-except para capturar a exceção e repetir a solicitação indefinidamente até que uma entrada válida seja fornecida, impedindo que o programa seja interrompido.

```python
# 7 Questão
 # Questão 7: Contagem crescente
print("--- Questão 7: Contagem crescente ---")

while True:
    try:
        # Tenta ler N e converter para inteiro
        N = int(input("Digite um número inteiro positivo N: "))

        # Valida se N é maior que 0
        if N <= 0:
            print("Erro: O número precisa ser maior que zero (N > 0). Tente novamente.")
        else:
            break # Entrada válida, sai do loop
    except ValueError:
        # Se o usuário digitar texto ou caracteres inválidos
        print("Erro: Entrada inválida! Digite um número inteiro.")

# Exibindo a contagem de 1 até N
print(f"Contagem de 1 até {N}:")
for i in range(1, N + 1):
    print(i, end=" ")
print() # Quebra de linha final
```

Questão 7
  
➥Exceção (ValueError): É disparada quando a função int() tenta converter um valor digitado pelo usuário que não representa um número inteiro válido (como letras ou símbolos).

➥​Estratégia: Utiliza a estrutura de repetição while True para forçar a entrada de dados correta. O bloco try/except captura valores não numéricos, enquanto a condição if N <= 0 impede números nulos ou negativos. O laço só é interrompido com break após o fornecimento de uma entrada válida, garantindo a execução do loop de contagem (for i in range(1, N + 1)).

```python
# 8 questão:

while True:
    try:
        # Tenta ler o valor e converter para inteiro
        n = int(input("Digite um número inteiro positivo para N: "))

        # Rejeita se N for menor ou igual a zero
        if n <= 0:
            print("Erro: O número precisa ser maior que zero. Tente novamente.")
            continue
        break # Sai do loop se a entrada for válida
    except ValueError:
        print("Erro: Digite apenas números inteiros válidos.")

# Lógica original: somar de 1 até N com acumulador
soma = 0
for i in range(1, n + 1):
    soma += i

print(f"A soma dos números de 1 até {n} é: {soma}")
```

 Questão 8


➥Exceção (ValueError): Tratada pelo bloco except ValueError, essa exceção captura situações em que o usuário digita um texto, letras ou caracteres que não podem ser convertidos para um número inteiro pela função int(), impedindo que o programa pare de funcionar com um erro.

➥​Estratégia: Utiliza um loop infinito (while True) como estrutura de validação contínua. O programa tenta ler o dado no bloco try; se houver erro de tipo, o except trata a exceção, e se o número for menor ou igual a zero (n <= 0), o continue força uma nova tentativa. O laço só é encerrado (break) quando uma entrada inteira e positiva é inserida.

```python
# 9 questão:
contador_pares = 0
quantidade_lidos = 0

# Repete até conseguir ler exatamente 10 valores válidos
while quantidade_lidos < 10:
    try:
        # Solicita o número da posição atual
        entrada = input(f"Digite o {quantidade_lidos + 1}º número inteiro: ")
        numero = int(entrada)

        # Se converteu com sucesso, verifica se é par
        if numero % 2 == 0:
            contador_pares += 1

        # Incrementa apenas se a entrada for válida
        quantidade_lidos += 1

    except ValueError:
        # Tratamento de exceção: não altera a quantidade lida, repete a posição
        print("Erro: Entrada inválida. Digite um número inteiro.")

print(f"Total de números pares digitados: {contador_pares}")
```

Questão 9​

➦Exceção (ValueError): Ocorre quando o usuário digita um valor inválido (como texto ou símbolos) que não pode ser convertido para número inteiro pela função int(entrada).  

​➥Estratégia: Utiliza o bloco try-except dentro de um laço while. O contador quantidade_lidos só é incrementado quando a conversão tem sucesso. Em caso de erro, a exceção exibe um aviso e o laço repete sem somar o ponto, garantindo a leitura correta de exatamente 10 números

```python
# 10 questão:
soma = 0
contador = 0

print("Digite 8 números inteiros:")
while contador < 8:
    try:
        # Lê o valor e tenta converter para inteiro
        numero = int(input(f"Digite o {contador + 1}º número: "))

        # Se for positivo, adiciona à soma
        if numero > 0:
            soma += numero

        # Incrementa o contador apenas se a entrada foi válida
        contador += 1
    except ValueError:
        # Tratamento caso o usuário digite algo que não seja inteiro (ex: letras, decimais)
        print("Entrada inválida! Por favor, digite um número inteiro.")

print(f"\nA soma dos números positivos é: {soma}")
```

​ValueError: Ocorre quando o usuário digita um valor que não pode ser convertido para inteiro pela função int() (por exemplo: letras, símbolos ou números decimais).
➥
​Estratégia:
​Validação sem interrupção: O bloco try...except impede que a aplicação feche inesperadamente ao receber dados inválidos, exibindo uma mensagem de aviso.
​Repetição garantida: O comando contador += 1 fica dentro do bloco try após a conversão. Se houver erro, a execução pula direto para o except, evitando que a tentativa inválida seja somada ao contador e forçando o laço a repetir até completar as 8 entradas válidas.

```python
# 11 questão:
maior = None
contador = 0

print("Digite 6 números inteiros:")
while contador < 6:
    try:
        numero = int(input(f"Digite o {contador + 1}º número: "))

        # Define o primeiro número válido como o maior inicial,
        # ou atualiza se encontrar um número maior
        if maior is None or numero > maior:
            maior = numero

        contador += 1
    except ValueError:
        print("Entrada inválida! Por favor, digite um número inteiro.")

print(f"\nO maior valor digitado foi: {maior}")
```

➥Estratégia: Utiliza um laço de repetição infinito (while True) que solicita a senha continuamente até que ela seja igual a "1234", momento em que o comando break encerra o programa com acesso permitido.

➥​Exceção: Ao manter a entrada do input() diretamente como texto (string), o código evita exceções de conversão (como ValueError, que ocorreria se tentasse converter a entrada para int). Qualquer sequência de caracteres ou símbolos digitada pelo usuário é processada com segurança sem travar o programa.

```python
# 14 questão:

numero_secreto = 7
tentativas = 0
acertou = False

print("Tente adivinhar o número secreto (entre 1 e 10).")

while not acertou:
  try:
    # No Google Colab, o input abre uma caixinha para digitar
    chute = int(input("Digite seu palpite: "))
    tentativas += 1

    if chute == numero_secreto:
      print(f"Parabéns! Você acertou em {tentativas} tentativas.")
      acertou = True
    elif chute < numero_secreto:
      print("O número secreto é maior.")
    else:
      print("O número secreto é menor.")

  except ValueError:
    # Tratamento para quando o usuário digita letras ou símbolos
    print("Entrada inválida! Digite apenas números inteiros.")
    tentativas += 1  # Conta a tentativa inválida conforme a regra da questão
```

Questão 14​

➥Exceção (ValueError): Ocorre quando o usuário digita um valor que não pode ser convertido para número inteiro (como texto ou símbolos) na função int().

​➥Estratégia: Utiliza o bloco try...except dentro do laço while para capturar o erro de digitação sem interromper o programa, informando a entrada inválida e permitindo que o usuário continue tentando até acer

```python
# 15 questão:
# Questão 15 - Analisando uma sequência
quantidade = 0
positivos = 0
negativos = 0
soma = 0

print("Digite números inteiros (digite 0 para encerrar):")

while True:
  try:
    entrada = input("Digite um número: ")
    numero = int(entrada)

    # Condição de parada
    if numero == 0:
      break

    # Atualiza as estatísticas
    quantidade += 1
    soma += numero

    if numero > 0:
      positivos += 1
    else:
      negativos += 1

  except ValueError:
    # Se digitar texto, o programa avisa e continua a execução sem quebrar
    print("Entrada inválida! Tente novamente com um número inteiro.")

# Exibição dos resultados finais
print("\n--- Resultado Final ---")
print(f"Quantidade de números válidos: {quantidade}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma total: {soma}")
```
