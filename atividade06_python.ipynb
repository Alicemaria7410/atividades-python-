```python
# 1 Questão

def ler_inteiro():
    while True:
        try:
            return int(input("Digite um número: "))
        except ValueError:
            print("Entrada inválida. Digite um inteiro.")

primeiro = ler_inteiro()
maior = primeiro
menor = primeiro

for i in range(5):
    numero = ler_inteiro()
    if numero > maior:
        maior = numero
    if numero < menor:
        menor = numero

print("Maior:", maior)
print("Menor:", menor)
```

```python
# 2 questão

soma = 0
aprovados = 0
abaixo = 0

for i in range(5):
    while True:
        try:
            nota = float(input(f"Nota do aluno {i+1}: "))
            if not 0 <= nota <= 10:
                print("Nota deve estar entre 0 e 10.")
                continue
            break
        except ValueError:
            print("Digite uma nota numérica válida.")

    soma += nota
    if nota >= 7:
        aprovados += 1
    else:
        abaixo += 1

print("Média:", soma / 5)
print("Notas >= 7:", aprovados)
print("Notas < 7:", abaixo)
```

Questão 06

➥​Estratégia: Utiliza o laço while True com try/except para repetir a solicitação até que a entrada seja válida (break). Isso garante que o laço for receba 8 valores inteiros válidos para encontrar e atualizar o maior número e sua primeira posição.

```python
# 7 questão
maior = None
segundo_maior = None
qtd_lidos = 0

# Para garantir números distintos sem usar listas, guardamos os números anteriores em variáveis simples se necessário,
# mas para ler 8 números únicos, validamos contra os maiores atuais ou comparamos.

# Forma simplificada de garantir 8 números únicos:
while qtd_lidos < 8:
    try:
        num = int(input(f"Digite o {qtd_lidos + 1}º número inteiro: "))
    except ValueError:
        print("Entrada inválida! Digite apenas números inteiros.")
        continue

    # Atualiza maior e segundo maior
    if maior is None:
        maior = num
        qtd_lidos += 1
    elif num == maior or (segundo_maior is not None and num == segundo_maior):
        print("Número repetido! Digite apenas números distintos.")
    else:
        if num > maior:
            segundo_maior = maior
            maior = num
            qtd_lidos += 1
        elif segundo_maior is None or num > segundo_maior:
            segundo_maior = num
            qtd_lidos += 1

print(f"\nMaior valor: {maior}")
print(f"Segundo maior valor: {segundo_maior}")
```

```python
# 8 questão
anterior = None
tem_repeticao = False

for i in range(1, 9):
    while True:
        try:
            num = int(input(f"Digite o {i}º número: "))
            break
        except ValueError:
            print("Entrada inválida! Digite um número inteiro.")

    if anterior is not None and num == anterior:
        tem_repeticao = True

    anterior = num

if tem_repeticao:
    print("\nExistem valores consecutivos iguais.")
else:
    print("\nNão existem valores consecutivos iguais.")
```

Questão 08

➥Exceção (ValueError): Captura o erro gerado quando a entrada do usuário não pode ser convertida para um número inteiro (como letras ou símbolos), impedindo que o programa pare de funcionar acidentalmente.  

➥​Estratégia: Usa um laço while True com try/except para insistir na leitura até que um número válido seja digitado. Além disso, armazena o número anterior na variável anterior a cada iteração para verificar se o valor atual é igual ao digitado imediatamente antes, sinalizando a repetição na variável tem_repeticao.

```python
# 9 questão
saldo = 500.0

for i in range(1, 7):
    print(f"\n--- Operação {i} (Saldo atual: R$ {saldo:.2f}) ---")

    # Valida o tipo de operação
    while True:
        try:
            tipo = int(input("Digite o tipo (1 = Depósito, 2 = Saque): "))
            if tipo in [1, 2]:
                break
            print("Código inválido! Escolha 1 para Depósito ou 2 para Saque.")
        except ValueError:
            print("Entrada inválida! Digite 1 ou 2.")

    # Valida o valor da operação
    while True:
        try:
            valor = float(input("Digite o valor da operação: R$ "))
            if valor > 0:
                break
            print("O valor deve ser positivo!")
        except ValueError:
            print("Entrada inválida! Digite um valor numérico.")

    # Executa a operação
    if tipo == 1:
        saldo += valor
        print(f"Depósito realizado com sucesso!")
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
            print(f"Saque realizado com sucesso!")
        else:
            print("Saldo insuficiente!")

print(f"\n========================")
print(f"Saldo final: R$ {saldo:.2f}")
```

Questão 9

➥Exceção: quando o usuário digita um valor de texto ou caractere incompatível que não pode ser convertido para número (int no tipo de operação ou float no valor).

​➥Estratégia: Utiliza a combinação de while True com try-except para forçar a repetição da leitura. O código só encerra o laço com o break quando o usuário fornece um dado do tipo correto e válido, evitando que o programa encerre por erro.

```python

# 10 questão
# Entrada com validação
while True:
    try:
        valor = int(input("Digite um valor inteiro positivo: "))
        if valor > 0:
            break
        print("Erro: Digite um número maior que zero.")
    except ValueError:
        print("Erro: Digite apenas números inteiros válidos.")

# Processamento e Saída
print(f"\nValor a ser sacado: R$ {valor}")
cedulas = [100, 50, 20, 10, 5, 2, 1]

for c in cedulas:
    qnt = valor // c
    valor %= c
    if qnt > 0:
        print(f"{qnt} nota(s) de R$ {c}")
```

➥​Estratégia:

```python
#12 questão
# Entrada com validação
while True:
    try:
        n = int(input("Digite um inteiro positivo N: "))
        if n > 0:
            break
        print("Erro: O número deve ser maior que zero.")
    except ValueError:
        print("Erro: Digite apenas números inteiros válidos.")

# Processamento
divisores = []
for i in range(1, n + 1):
    if n % i == 0:
        divisores.append(i)

# Saída
print(f"Divisores de {n}: {divisores}")
print(f"Total de divisores: {len(divisores)}")
```

Questão 12

➥Exceção (ValueError): É acionada se o usuário digitar textos, letras ou números decimais que o int() não consegue converter.

➥​Estratégia de validação: O laço while True repete a solicitação até que uma entrada válida e maior que zero (n > 0) seja fornecida, interrompendo a repetição com o break.

```python
#13 questão
# Entrada com validação
while True:
    try:
        a = int(input("Digite A (positivo): "))
        b = int(input("Digite B (positivo, maior que A): "))
        if a > 0 and b > a:
            break
        print("Erro: A e B devem ser positivos e A deve ser menor que B.")
    except ValueError:
        print("Erro: Digite apenas números inteiros válidos.")

# Processamento
primos = []
for num in range(a, b + 1):
    if num > 1:
        eh_primo = True
        for i in range(2, num):
            if num % i == 0:
                eh_primo = False
                break
        if eh_primo:
            primos.append(num)

# Saída
print(f"Números primos entre {a} e {b}: {primos}")
print(f"Total de primos encontrados: {len(primos)}")
```

Questão 13

➥Exceção (ValueError): Trata o erro gerado caso o usuário digite texto, letras ou símbolos em vez de números inteiros válidos ao executar o int(input(...)).

➥​Estratégia:
​Validação de entrada: Usa um laço infinito (while True) com o bloco try/except para garantir que o programa só avança quando a e b forem números inteiros válidos, positivos e com b > a.
​Verificação de números primos: Percorre o intervalo de a até b e testa se cada número possui divisores entre 2 e num - 1. Se nenhum divisor for encontrado, o número é primo e é adicionado à lista primos.

```python
#14 questão
SECRETO = 37
tentativas = 0

while tentativas < 5:
    try:
        palpite = int(input(f"Tentativa {tentativas + 1}/5 - Digite seu palpite: "))
    except ValueError:
        print("Texto/Valor inválido! Tente digitar um número inteiro.")
        continue  # Não gasta tentativa se for texto inválido

    tentativas += 1

    if palpite == SECRETO:
        print(f"Parabéns! Você acertou o número secreto ({SECRETO}) em {tentativas} tentativa(s)!")
        break
    elif palpite < SECRETO:
        print("O número secreto é MAIOR que o seu palpite.")
    else:
        print("O número secreto é MENOR que o seu palpite.")

if palpite != SECRETO and tentativas == 5:
    print(f"\nGame Over! Suas 5 tentativas acabaram. O número secreto era {SECRETO}.")
```

Questão 14​

➥Exceção (ValueError): Captura erros de conversão quando o usuário digita textos, letras ou caracteres que não podem ser transformados em número inteiro pela função int().

➥​Estratégia: Utiliza a instrução continue no bloco except para reiniciar o laço imediatamente, pulando a linha tentativas += 1. Com isso, entradas inválidas exibem uma mensagem de alerta sem gastar o limite de tentativas do jogador.

```python
# 15 questão
saldo = 200.0
qtd_entradas = 0
qtd_retiradas = 0
retiradas_recusadas = 0
total_entradas = 0.0
total_saidas = 0.0

while True:
    # 1. Leitura e validação do CÓDIGO de operação
    try:
        codigo = int(input("Informe o código (1 = Entrada, 2 = Retirada, 0 = Sair): "))
    except ValueError:
        print("Erro: Digite apenas números inteiros!")
        continue

    if codigo == 0:
        break  # Encerra o programa
    elif codigo not in [1, 2]:
        print("Código inválido! Tente novamente.")
        continue

    # 2. Leitura e validação do VALOR da operação
    try:
        valor = float(input("Informe o valor: "))
    except ValueError:
        print("Erro: Digite um valor numérico válido!")
        continue

    if valor <= 0:
        print("O valor deve ser maior que zero!")
        continue

    # 3. Processamento das Operações
    if codigo == 1:
        saldo += valor
        total_entradas += valor
        qtd_entradas += 1
        print("Entrada realizada com sucesso!")

    elif codigo == 2:
        if valor <= saldo:
            saldo -= valor
            total_saidas += valor
            qtd_retiradas += 1
            print("Retirada realizada com sucesso!")
        else:
            retiradas_recusadas += 1
            print("Retirada RECUSADA: Saldo insuficiente!")

# 4. Exibição dos resultados finais
print("\n--- RESUMO DO CAIXA ---")
print(f"Saldo final: R$ {saldo:.2f}")
print(f"Quantidade de entradas: {qtd_entradas}")
print(f"Quantidade de retiradas realizadas: {qtd_retiradas}")
print(f"Quantidade de retiradas recusadas: {retiradas_recusadas}")
print(f"Total de dinheiro que entrou: R$ {total_entradas:.2f}")
print(f"Total de dinheiro que saiu: R$ {total_saidas:.2f}")
```
